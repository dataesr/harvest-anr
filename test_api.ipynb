```python
import requests
import json

url_search = "https://api.tech.ec.europa.eu/search-api/prod/rest/search"
url_facet = "https://api.tech.ec.europa.eu/search-api/prod/rest/facet"

headers = {"Accept": "application/json, text/plain, */*"}
```

```python
query = {
    "bool": {
        "must": [
            {"terms": {"DATASOURCE": ["SEDIA_NONH2020_PROD"]}},
            {"terms": {"status": ["Ongoing"]}},
            {"terms": {"language": ["en"]}},
        ],
        "must_not": [{"terms": {"programId": ["43108390", "31045243"]}}],
    }
}
sort = {"order": "DESC", "field": "es_SortDate"}
fields = [
    "title",
    "acronym",
    "programId",
    "projectId",
    "callIdentifier",
    "programAbbreviation",
    "programmes",
    "status",
    "objectives",
    "participants",
    "topicAbbreviation",
    "topicDescription",
    "overallBudget",
    "euContributionRate",
    "euContributionAmount",
    "freeKeywords",
]
languages = ["en"]

files = {
    "query": ("blob", json.dumps(query), "application/json"),
    "sort": ("blob", json.dumps(sort), "application/json"),
    # "displayFields": ("blob", json.dumps(fields), "application/json"),
    "languages": ("blob", json.dumps(languages), "application/json"),
}

params = {"apiKey": "SEDIA_NONH2020_PROD", "text": "***", "pageSize": 1, "pageNumber": 1}
```

```python
res = requests.post(url=url_search, headers=headers, params=params, files=files)
res.json()
```

```python
metadata = res.json().get("results", [])[0].get("metadata")
participants = json.loads(metadata.get("participants")[0])
```
