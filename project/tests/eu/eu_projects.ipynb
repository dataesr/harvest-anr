## TEST API

```python
import requests
import json

url_search = "https://api.tech.ec.europa.eu/search-api/prod/rest/search"
url_facet = "https://api.tech.ec.europa.eu/search-api/prod/rest/facet"

headers = {"Accept": "application/json, text/plain, */*"}
```

```python
query = {
    "bool": {
        "must": [
            {"terms": {"DATASOURCE": ["SEDIA_NONH2020_PROD"]}},
            {"terms": {"status": ["Ongoing"]}},
            {"terms": {"language": ["en"]}},
        ],
        "must_not": [{"terms": {"programId": ["43108390", "31045243"]}}],
    }
}
sort = {"order": "DESC", "field": "es_SortDate"}
fields = [
    "title",
    "acronym",
    "programId",
    "projectId",
    "callIdentifier",
    "programAbbreviation",
    "programmes",
    "status",
    "objectives",
    "participants",
    "topicAbbreviation",
    "topicDescription",
    "overallBudget",
    "euContributionRate",
    "euContributionAmount",
    "freeKeywords",
]
languages = ["en"]

files = {
    "query": ("blob", json.dumps(query), "application/json"),
    "sort": ("blob", json.dumps(sort), "application/json"),
    # "displayFields": ("blob", json.dumps(fields), "application/json"),
    "languages": ("blob", json.dumps(languages), "application/json"),
}

params = {"apiKey": "SEDIA_NONH2020_PROD", "text": "***", "pageSize": 1, "pageNumber": 1}
```

```python
res = requests.post(url=url_search, headers=headers, params=params, files=files)
res.json()
```

```python
metadata = res.json().get("results", [])[0].get("metadata")
participants = json.loads(metadata.get("participants")[0])
```

## TEST HARVEST

```python
import pandas as pd

# run eu_projects.py first
projects_df = pd.read_json("eu_projects.jsonl", lines=True)
projects_df.head()
```

```python
projects_df["type"].value_counts()
```

```python
def flatten(xss):
    return [x for xs in xss for x in xs]
```

```python
from project.server.main.utils import normalize

participants_df = pd.DataFrame.from_records(flatten(projects_df["participants"].to_list()))
participants_df["label"] = participants_df["label"].apply(lambda x: x.get("default"))
participants_df["label_norm"] = participants_df["label"].apply(lambda x: normalize(x))
```

```python
participants_df.head()
```

```python
unique_df = participants_df.groupby("label_norm", as_index=False).agg(
    **{col: (col, "first") for col in participants_df.columns if col != "label_norm"}, count=("label_norm", "size")
)
unique_df.head()
```

```python
cache_df = pd.read_json("../../../cache_participant.json", orient="index").reset_index()
cache_df = cache_df.rename(columns={"index": "label_norm", 0: "cache_id"})
cache_df
```

```python
unique_df = unique_df.merge(cache_df, how="left", on="label_norm")
unique_df
```

```python
unique_df["cache_id"].notna().sum()
```

```python
unique_df = unique_df.sort_values("count", ascending=False)
```

```python
found = unique_df["cache_id"].notna().sum()
not_found = unique_df.shape[0] - found
print("found", found)
print("not_found", not_found)
```

```python
unique_df[~unique_df["cache_id"].notna()][["label", "label_norm", "count"]].to_csv(
    "eu_part_to_identify.csv", encoding="utf-8"
)
```
